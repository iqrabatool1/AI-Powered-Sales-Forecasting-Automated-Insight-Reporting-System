```python

# Import required libraries

import os
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

from PIL import Image
import hashlib

print("All libraries imported successfully!")
```

```python

# Define the main dataset path

BASE_DIR = "/content/drive/MyDrive/IDRID/IDRiD"

# Training dataset paths
TRAIN_DIR = os.path.join(BASE_DIR, "Train", "images")
TRAIN_CSV = os.path.join(BASE_DIR, "Train", "annotations.csv")

# Testing dataset paths
TEST_DIR = os.path.join(BASE_DIR, "Test", "images")
TEST_CSV = os.path.join(BASE_DIR, "Test", "annotations.csv")

print("Dataset paths defined successfully!")
```

```python

# Check whether dataset folders and files exist

print("Training images folder:", os.path.exists(TRAIN_DIR))
print("Training annotations CSV:", os.path.exists(TRAIN_CSV))

print("Testing images folder:", os.path.exists(TEST_DIR))
print("Testing annotations CSV:", os.path.exists(TEST_CSV))
```

```python

# Load training and testing annotation files

train_df = pd.read_csv(TRAIN_CSV)
test_df = pd.read_csv(TEST_CSV)

# Display the first five records
print("TRAINING DATA:")
display(train_df.head())

print("\nTESTING DATA:")
display(test_df.head())
```

```python

# Select only the required columns

train_df = train_df[["Image name", "Retinopathy grade"]].copy()
test_df = test_df[["Image name", "Retinopathy grade"]].copy()

# Rename columns
train_df.columns = ["filename", "label"]
test_df.columns = ["filename", "label"]

# Display prepared training data
print("Prepared Training Data:")
display(train_df.head())

print("\nPrepared Testing Data:")
display(test_df.head())
```

```python

# Count training and testing images

print("Total Training Images:", len(train_df))
print("Total Testing Images:", len(test_df))

# Count images in each severity grade
print("\nTraining Class Distribution:")
print(train_df["label"].value_counts().sort_index())

print("\nTesting Class Distribution:")
print(test_df["label"].value_counts().sort_index())
```

```python

# Check whether all training and testing images can be opened

from PIL import Image

def check_images(image_dir, dataframe):
    valid_images = 0
    invalid_images = []

    for filename in dataframe["filename"]:
        image_path = os.path.join(image_dir, filename)

        try:
            with Image.open(image_path) as img:
                img.verify()
            valid_images += 1

        except Exception:
            invalid_images.append(filename)

    print("Total images:", len(dataframe))
    print("Valid images:", valid_images)
    print("Invalid images:", len(invalid_images))

    if invalid_images:
        print("Invalid filenames:", invalid_images[:10])

print("TRAINING IMAGE CHECK")
check_images(TRAIN_DIR, train_df)

print("\nTESTING IMAGE CHECK")
check_images(TEST_DIR, test_df)
```
